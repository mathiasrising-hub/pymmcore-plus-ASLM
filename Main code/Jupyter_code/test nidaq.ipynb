```python
import nidaqmx
import numpy as np
import warnings
import time
warnings.simplefilter("ignore", nidaqmx.errors.DaqResourceWarning)
```

```python
task_co = nidaqmx.Task()
task_co.co_channels.add_co_pulse_chan_freq('Dev1/ctr0', name_to_assign_to_channel='pulse_gen', freq=2, duty_cycle=0.1)
task_co.timing.cfg_implicit_timing(nidaqmx.constants.AcquisitionType.CONTINUOUS)


arr = np.array([[False, True], [True,False]],dtype = np.bool_)
task_do = nidaqmx.Task()
task_do.do_channels.add_do_chan('Dev1/port0/line23,Dev1/port0/line24',line_grouping=nidaqmx.constants.LineGrouping.CHAN_PER_LINE)
task_do.timing.cfg_samp_clk_timing(1,
                                   source = '/Dev1/Ctr0InternalOutput',
                                    sample_mode=nidaqmx.constants.AcquisitionType.FINITE
                                    )
task_do.write(arr, False)

task_co.start()
task_do.start()

time.sleep(10)

task_co.stop()
task_do.stop()
```

```python
task_co.close()
task_do.close()
```
